# Exploratory Data Analysis: Cardiotocographic Dataset


### Objective
The objective of this assignment is to perform exploratory data analysis on the Cardiotocographic dataset and understand the structure, distribution, relationships, and possible outliers in the data.

### Methodology
The dataset is loaded and inspected using data types, missing values, and summary statistics. Histograms are used to understand feature distributions, boxplots are used to detect outliers, and a correlation heatmap is used to identify relationships between variables.

### Interpretation
EDA helps identify data quality issues and important patterns before modeling. Strong correlations may show related medical measurements, while outliers may represent unusual patient records or values that need further investigation.


```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('Cardiotocographic.csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.hist(figsize=(20,15))
plt.show()
```

```python
plt.figure(figsize=(20,15))
sns.boxplot(df)
plt.show()
```

```python
df.skew()
```

```python
sns.scatterplot(x='LB', y='AC', data=df)
plt.show()
```

```python
plt.figure(figsize=(20,15))
sns.pairplot(df)
plt.show()
```

```python
plt.figure(figsize=(20,15))
sns.heatmap(df.corr(),annot=True)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
df['NSP'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

# Interview Questions

## 1. What is Exploratory Data Analysis (EDA)?
EDA is the process of analyzing datasets using statistical and visualization techniques to understand patterns, relationships, and anomalies.

## 2. Why is data cleaning important?
Data cleaning improves data quality by handling missing values, duplicates, and inconsistencies.

## 3. What are outliers?
Outliers are extreme values that differ significantly from other observations.

## 4. What is correlation?
Correlation measures the relationship between two variables.

## 5. Why is a heatmap used?
Heatmaps help visualize correlations between numerical variables.

## 6. Difference between histogram and boxplot?
Histogram shows distribution frequency, while boxplot shows spread and outliers.

## 7. What is IQR?
Interquartile Range measures statistical dispersion and is used for outlier detection.

## 8. Why use pairplot?
Pairplot helps visualize relationships among multiple variables simultaneously.

## 9. What are missing values?
Missing values are absent or null entries in the dataset.

## 10. Why is visualization important in EDA?
Visualization helps identify trends, patterns, and anomalies quickly.
