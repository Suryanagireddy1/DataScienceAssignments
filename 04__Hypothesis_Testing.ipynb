# Hypothesis Testing

## Assignment Explanation

### Objective
The objective of this assignment is to apply statistical hypothesis testing to business problems and make decisions using test statistics and p-values.

### Methodology
The first test checks whether the observed weekly operating cost is significantly higher than the theoretical cost model. A one-tailed z-test is used because the population standard deviation can be derived from the given cost formula.

The second test uses a Chi-Square test of independence to check whether customer satisfaction is associated with the type of smart-home device purchased.

### Interpretation
In hypothesis testing, the null hypothesis represents no change or no association. If the calculated p-value is less than the significance level, the null hypothesis is rejected. This means there is enough statistical evidence to support the alternative claim.


```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Operating Cost Test


```python
sample_mean = 3050
mu_x = 600
sigma_x = 25
n = 25
theoretical_mean = 1000 + 5 * mu_x
sigma_w = 5 * sigma_x

z = (sample_mean - theoretical_mean) / (sigma_w / np.sqrt(n))
p_value = 1 - stats.norm.cdf(z)
z, p_value
```

```python
alpha = 0.05
decision = 'Reject H0' if p_value < alpha else 'Fail to reject H0'
decision
```

## Chi-Square Test for Device Type and Satisfaction


```python
observed = pd.DataFrame({
    'Smart Thermostat': [50, 80, 60, 30, 20],
    'Smart Light': [70, 100, 90, 50, 50]
}, index=['Very Satisfied', 'Satisfied', 'Neutral', 'Unsatisfied', 'Very Unsatisfied'])
chi2, p, dof, expected = stats.chi2_contingency(observed)
chi2, p, dof
```

```python
critical = stats.chi2.ppf(0.95, dof)
decision = 'Reject H0: association exists' if chi2 > critical else 'Fail to reject H0: no significant association'
critical, decision
```
