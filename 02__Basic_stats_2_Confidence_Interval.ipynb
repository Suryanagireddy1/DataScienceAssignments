# Estimation and Confidence Intervals


## Assignment Explanation

### Objective
The objective of this assignment is to estimate the average durability of print-heads using confidence intervals.

### Methodology
A sample of 15 print-head durability values is used. Since the sample size is small, the confidence interval is first calculated using the sample standard deviation and the t-distribution. A second confidence interval is calculated using the known population standard deviation and the z-distribution.

### Interpretation
The t-based interval is wider because the population standard deviation is unknown and the estimate comes from a small sample. The z-based interval is narrower because it assumes the population standard deviation is already known. Both intervals give a possible range for the true average durability of print-heads.


```python
import numpy as np
from scipy import stats

data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
n = len(data)
mean = data.mean()
sample_std = data.std(ddof=1)
population_std = 0.2
mean, sample_std, n
```

```python
# 99% CI using sample standard deviation: t interval
alpha = 0.01
t_crit = stats.t.ppf(1 - alpha/2, df=n-1)
margin_t = t_crit * sample_std / np.sqrt(n)
ci_t = (mean - margin_t, mean + margin_t)
ci_t
```

```python
# 99% CI using known population standard deviation: z interval
z_crit = stats.norm.ppf(1 - alpha/2)
margin_z = z_crit * population_std / np.sqrt(n)
ci_z = (mean - margin_z, mean + margin_z)
ci_z
```

## Conclusion

The t interval is wider because the population standard deviation is unknown and must be estimated from a small sample.
