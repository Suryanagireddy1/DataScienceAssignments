## Exercise 1: Prime Numbers

A prime number is greater than 1 and has no divisors other than 1 and itself. The function below checks all possible divisors up to the square root of the number.


```python
def is_prime(number):
    if number <= 1:
        return False

    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False

    return True

test_number = 29
print(f'{test_number} is prime:', is_prime(test_number))
```

## Exercise 2: Product of Random Numbers

This exercise generates two random numbers, calculates their product, and checks whether the entered answer is correct. A sample answer is used so the notebook can run without waiting for keyboard input.


```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)
correct_product = num1 * num2

# Replace this value with input() if running interactively.
user_answer = correct_product

print(f'Number 1: {num1}')
print(f'Number 2: {num2}')
print(f'Your answer: {user_answer}')

if user_answer == correct_product:
    print('Correct answer')
else:
    print(f'Incorrect answer. The correct product is {correct_product}')
```

## Exercise 3: Squares of Even Numbers

The assignment asks to choose either even or odd numbers between 100 and 200. Here, even numbers are selected and their squares are printed.


```python
even_squares = {number: number ** 2 for number in range(100, 201) if number % 2 == 0}

for number, square in even_squares.items():
    print(f'{number} squared = {square}')
```

## Exercise 4: Word Counter

This program counts how many times each word appears in a given text. The text is converted to lowercase so words such as This and 	his are counted together.


```python
from collections import Counter
import re

input_text = 'This is a sample text. This text will be used to demonstrate the word counter.'

words = re.findall(r'\\b\\w+\\b', input_text.lower())
word_counts = Counter(words)

for word, count in word_counts.items():
    print(f'{word}: {count}')
```

## Exercise 5: Check for Palindrome

A palindrome reads the same forward and backward. The function below removes spaces, punctuation, and case differences before checking the string.


```python
def is_palindrome(text):
    cleaned_text = ''.join(character.lower() for character in text if character.isalnum())
    return cleaned_text == cleaned_text[::-1]

examples = ['madam', 'racecar', 'hello', 'A man, a plan, a canal: Panama']

for example in examples:
    print(f'{example}: {is_palindrome(example)}')
```
