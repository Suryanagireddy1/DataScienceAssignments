# Artificial Neural Network Classification

## Objective
The objective of this assignment is to classify alphabet characters using Artificial Neural Networks and improve model performance through hyperparameter tuning.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

```python
df = pd.read_csv("/content/Alphabets_data.csv")
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
X = df.drop('letter', axis=1)
y = df['letter']
```

```python
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
sc = StandardScaler()
X_scaled = sc.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = Sequential()

model.add(Dense(64, activation='relu', input_dim=X_train.shape[1]))
model.add(Dense(32, activation='relu'))
model.add(Dense(26, activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2
)
```

```python
y_pred = model.predict(X_test)
y_pred = np.argmax(y_pred, axis=1)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
tuned_model = Sequential()

tuned_model.add(Dense(128, activation='relu', input_dim=X_train.shape[1]))
tuned_model.add(Dense(64, activation='relu'))
tuned_model.add(Dense(32, activation='relu'))
tuned_model.add(Dense(26, activation='softmax'))
```

```python
optimizer = Adam(learning_rate=0.001)

tuned_model.compile(
    optimizer=optimizer,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history2 = tuned_model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2
)
```

```python
y_pred_tuned = tuned_model.predict(X_test)
y_pred_tuned = np.argmax(y_pred_tuned, axis=1)

print("Accuracy:", accuracy_score(y_test, y_pred_tuned))
print(classification_report(y_test, y_pred_tuned))
```

# Interview Questions and Answers

## 1. What is an Artificial Neural Network (ANN)?

An Artificial Neural Network is a machine learning model inspired by the structure of the human brain. It consists of interconnected neurons that process and learn patterns from data.

---

## 2. Why are Neural Networks used?

Neural Networks are used to solve complex problems such as classification, prediction, image recognition, speech processing, and pattern detection.

---

## 3. What are the main components of an ANN?

The main components of an ANN are:
- Input Layer
- Hidden Layer(s)
- Output Layer
- Weights
- Bias
- Activation Functions

---

## 4. What is the role of the hidden layer in ANN?

Hidden layers help the network learn complex relationships and patterns from the input data.

---

## 5. What is an activation function?

An activation function determines whether a neuron should be activated or not by introducing non-linearity into the model.

---

## 7. Why is Softmax used in the output layer?

Softmax converts output values into probability distributions for multiclass classification problems.

---

## 8. What is backpropagation?

Backpropagation is the process of updating weights in the neural network by minimizing the error using gradient descent.

---

## 9. What is the purpose of feature scaling in ANN?

Feature scaling helps improve training speed and model performance by bringing all feature values into a similar range.

---

## 10. What is hyperparameter tuning?

Hyperparameter tuning is the process of optimizing parameters such as:
- Number of hidden layers
- Number of neurons
- Learning rate
- Batch size
- Activation functions

to improve model performance.

---

## 11. What is overfitting in Neural Networks?

Overfitting occurs when the model learns training data too well and performs poorly on unseen test data.

---

## 12. What is underfitting?

Underfitting occurs when the model fails to learn important patterns from the dataset.

---

## 13. What is the difference between Epoch and Batch Size?

- Epoch: One complete pass through the entire training dataset.
- Batch Size: Number of samples processed before updating weights.

---

## 14. Why is Adam optimizer commonly used?

Adam optimizer combines the advantages of Momentum and RMSProp, providing faster and more efficient training.
